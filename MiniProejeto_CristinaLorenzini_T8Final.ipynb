# LIMPAR
Limpeza e decisões de Limpeza - pós diagnóstico inicial

* **Decisões:** Foram removidas duplicatas confirmadas, colunas vazias

**Funções**:

Criação de funções que automatizam o tratamento e o resumo dos dados.


### Remoção de Colunas Fantasmas
* **Problema:** O diagnóstico revelou a presença de quatro colunas (`Unnamed: 10` a `Unnamed: 13`) preenchidas inteiramente com valores nulos (`NaN`), geradas por delimitadores (`;`) extras na exportação do CSV original.
* **Decisão:** Excluir essas quatro colunas da base de trabalho (`df_limpo`) utilizando o método `.drop()`.
* **Motivo:** Essas colunas não possuem dados válidos e não fazem parte da unidade de análise (compras). Removê-las preserva apenas a informação útil, limpa a visualização da tabela e otimiza a memória do projeto.

### Tratamento de Inconsistência de Texto (Falsos Vazios)
* **Problema:** As colunas `Categoria_Produto` e `Nome_Produto` contêm o texto `'#N/D'`, que mascara a ausência de dados, impedindo o Python de contar corretamente os valores nulos.
* **Decisão:** Converter o texto `'#N/D'` para o valor nulo oficial do Python (`NaN`) e, em seguida, preencher esses espaços com `"Não Informado"`.
* **Motivo:** O Python lê `'#N/D'` como uma palavra comum. Ao transformar em "Não Informado", padronizamos a categoria e preservamos a linha na tabela, garantindo que o valor financeiro daquela compra não seja perdido.

### Remoção de Linhas Duplicadas
* **Problema:** A base pode conter registros inseridos em duplicidade acidental no sistema (linhas 100% idênticas).
* **Decisão:** Contar as duplicatas e utilizar o método `.drop_duplicates()` para remover apenas as repetições exatas[cite: 1, 4].
* **Motivo:** Removemos apenas as linhas 100% duplicadas, preservando compras distintas feitas pelo mesmo cliente[cite: 2]. Isso evita que contagens e faturamentos sejam inflados artificialmente pela duplicação.

````python
import numpy as np

# 1. CRIANDO A FUNÇÃO DE LIMPEZA
def tratar_base_varejo(df):
    df_limpo = df.copy()

    # Remoção das colunas fantasmas
    colunas_fantasmas = ['Unnamed: 10', 'Unnamed: 11', 'Unnamed: 12', 'Unnamed: 13']
    df_limpo = df_limpo.drop(columns=colunas_fantasmas, errors='ignore')

    # Tratamento de inconsistências (#N/D)
    df_limpo = df_limpo.replace('#N/D', np.nan)
    df_limpo = df_limpo.fillna("Não Informado")

    # Remoção de duplicatas exatas
    df_limpo = df_limpo.drop_duplicates()

    return df_limpo # Retorno obrigatório exigido na avaliação

# 2. APLICANDO A FUNÇÃO E FAZENDO A AUDITORIA
# Usamos 'df' (carregado com sep=';' na fase de diagnóstico) como nossa base de trabalho
linhas_brutas = len(df)

# Sobrescrevendo a variável df com a base limpa (Opção 1)
df = tratar_base_varejo(df)

linhas_limpas = len(df)
duplicatas_encontradas = linhas_brutas - linhas_limpas

print("✅ LIMPEZA COMPLETA - Resumo:\n")
print("-> Colunas fantasmas removidas.")
print("-> Falsos vazios convertidos para 'Não Informado'.")
print(f"-> Duplicatas exatas removidas: {duplicatas_encontradas}")
print("-" * 40)
print(f"Tamanho da base ANTES da limpeza: {linhas_brutas} linhas.")
print(f"Tamanho da base DEPOIS da limpeza (variável 'df'): {linhas_limpas} linhas válidas prontas para análise.")
````
